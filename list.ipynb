# LIST

```python
print("Operation of list")
animals=["cat", "dog", "monkey", "donkey","rohit"]
animals2=["omkar","monty","sahil"]

animals.append("eresh")
print (animals)

animals.insert(2,"sayee")
print (animals)

animals.remove("donkey")
print (animals)

animals.pop(3)
print (animals)

animals.extend(animals2)
print(animals)

animals.reverse()
print (animals)

a=animals.count("eresh")
print (a)

x=animals.copy()
print(x)

animals.sort()
print(animals)
```
