# While Loop

* 1)odd-even using while loop 

```python
number = 1
while number <= 20:
    if number % 2 == 0:
        print(f"{number} is even")
    else:
        print(f"{number} is odd")
    number += 1
```

* 2) pyramid pattern using while

```python
n = 6
i = 0

while i < n:
    print(' ' * (n - i + 1) + '*' * (i + 1))
    i += 1
```

* 3) creat list 1-20 numbers list using while loop=> [1,2,3... 20]

```python
numbers = []
i = 1
while i <= 20:
    numbers.append(i)
    i += 1
print(numbers)
```

* 4)  creat list 20-1 (revers order) using while loop=> [20,19...1]

```python
number = 20 
 
while number >= 1: 
    print(number) 
    number -= 1 
```

* 5) try with one any eg. break, continue , pass control statement

break

```python
i = 0
while i < 5:
    if i == 3:
        break 
    print(i)
    i += 1
```

continue

```python
for i in range(5):
    if i == 3:
        continue 
        
    print(i)
```

pass

```python
for i in range(5):
    if i == 3:
        pass 
        
    print(i)
```
