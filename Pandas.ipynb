* Aggregation and Grouping:

* Aggregation: .sum(), .mean(), .count()

* Merging and Joining:

* Condition Selection:
And ,or

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Sayee', 'Snehal', 'Gauri', 'Vaibhavi'],
    'Age': [28, 24, 35, 32],
    'Country': ['USA', 'UK', 'Australia', 'Germany']
})

selected_rows = df[(df['Age'] > 30) & (df['Country'] != 'USA')]

print(selected_rows)
df
```

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Sayee', 'Snehal', 'Gauri', 'Vaibhavi'],
    'Age': [28, 24, 35, 32],
    'Country': ['USA', 'UK', 'Australia', 'Germany']
})

selected_rows = df[(df['Age'] > 30) | (df['Country'] == 'USA')]

print(selected_rows)
df
```
