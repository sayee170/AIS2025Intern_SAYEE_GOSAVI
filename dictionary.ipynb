# DICTIONARY


```python
my_dict = {'name': 'SAYEE','age': 21,'city': 'Pune'}
x=my_dict.keys()
print(x)
y=my_dict.values()
print(y)
```

```python
my_dict = {'name': 'ABC','age': 20,'city': 'Pune'}

z=my_dict.items()
print(z)
q=my_dict.pop('age')
print(q)
```

```python
my_dict = {'name': 'ABC','age': 20,'city': 'Pune'}

my_dict.update({'bike':'hero'})
print(my_dict)
```
